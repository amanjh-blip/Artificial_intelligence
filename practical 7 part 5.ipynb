# IRIS Dataset part 5#

```python
from sklearn.datasets import load_iris
from sklearn.model_selection import cross_val_score
from sklearn.neighbors import KNeighborsClassifier
```

```python
#Load dataset
iris = load_iris()

x = iris.data
y = iris.target
```

```python
#create classification model
model = KNeighborsClassifier()
```

```python
#Apply 5-fold cross-validation
scores = cross_val_score(model, x, y, cv=5)

print("Accuracy scores for each fold:")
print(scores)

print("Average Accuracy:",round(scores.mean()*100, 2), "%")
```

# Employee_salary dataset #

```python
import pandas as pd
from sklearn.model_selection import cross_val_score
from sklearn.neighbors import KNeighborsClassifier
```

```python
# Load dataset
data = pd.read_csv("employee_salary.csv")
```

```python
# Create target class
data["Salary_Class"] = (data["Salary"] > data["Salary"].median()).astype(int)
```

```python
# Features and Target
X = data[["Age", "Years_of_Experience"]]
y = data["Salary_Class"]
```

```python
# Create Classification Model
model = KNeighborsClassifier()

# Apply 5-Fold Cross Validation
scores = cross_val_score(model, X, y, cv=5)

print("Accuracy scores for each fold:")
print(scores)

print("Average Accuracy:", round(scores.mean() * 100, 2), "%")
```
